## Text Parsing and Chunking

This section demonstrates text parsing and chunking using regular expressions and the spaCy library.

```python
import re
import spacy
```

```python
sample_text = (
    "The quick brown fox jumps over the lazy dog. "
    "It was a sunny day, and the birds were singing. "
    "The cat, a fluffy Persian, napped peacefully on the couch. "
    "John Doe, a renowned linguist, published a new paper on syntax. "
    "His research focuses on sentence structure and semantic roles."
    "The project received funding from the National Science Foundation (NSF)."
)
```

```python
def regex_chunking(text):
    sentence_chunks = re.findall(r'[^.!?]+[.!?]', text)
    noun_phrases = re.findall(r'\b(?:(?:[ADJ]* |[DET]* |[PRON]* )*(?:[NOUN]+ )+)\b', text, re.IGNORECASE)
    return sentence_chunks, noun_phrases

sentence_chunks_regex, noun_phrases_regex = regex_chunking(sample_text)

print("Sentence Chunks (Regex):")
for chunk in sentence_chunks_regex:
    print(f"- {chunk.strip()}")

print("\nNoun Phrases (Regex):")
for phrase in noun_phrases_regex:
    print(f"- {phrase.strip()}")
```

```python
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Downloading spaCy model 'en_core_web_sm'...")
    spacy.cli.download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")
```

```python
def spacy_chunking(text):
    doc = nlp(text)
    noun_chunks = [chunk.text for chunk in doc.noun_chunks]
    verb_phrases = []
    for token in doc:
        if "VERB" in token.pos_:
            start = token.i
            while start > 0 and doc[start - 1].dep_ in ("aux", "advmod", "neg", "prt"):
                start -= 1
            end = token.i + 1
            while end < len(doc) and doc[end].dep_ in ("dobj", "prep", "advmod", "prt", "acomp"):
                end += 1
            verb_phrases.append(doc[start:end].text)
    return noun_chunks, verb_phrases

noun_chunks_spacy, verb_phrases_spacy = spacy_chunking(sample_text)

print("Noun Chunks (spaCy):")
for chunk in noun_chunks_spacy:
    print(f"- {chunk.strip()}")

print("\nVerb Phrases (spaCy):")
for phrase in verb_phrases_spacy:
    print(f"- {phrase.strip()}")
```
