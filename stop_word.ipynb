### Stop Word Removal

Stop words are common words (like 'the', 'is', 'and') that often carry little meaning in text analysis and can be removed to focus on more significant terms. We'll use `nltk` for this task.

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    word_tokens = word_tokenize(text)
    filtered_text = [word for word in word_tokens if word.lower() not in stop_words]
    return " ".join(filtered_text)
```

```python
document = "This is an example sentence showing how to remove common words, also known as stop words, using the NLTK library in Python."

cleaned_document = remove_stopwords(document)
print("Original document:", document)
print("Cleaned document:", cleaned_document)
```
